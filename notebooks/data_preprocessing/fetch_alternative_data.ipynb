# Fetch Alternative Data

## Overview

This notebook saves timestamped news for later sentiment scoring.
- Problem: alternative signals require news linked to symbols and publication times.
- Approach: fetch Alpaca News API responses and store normalized articles as parquet.
- Alternative Data: It retains headlines, summaries, content, and article metadata for sentiment analysis.

```python
from datetime import datetime, timezone

import pandas as pd
from IPython.display import display

from src.data_preprocessing.fetch_alternative_data import save_alpaca_news
```

## Configure Request

This cell defines a narrow news request.
- `symbols` filters articles by related instruments.
- `start` and `end` define the UTC request interval.
- `output_path=None` uses the module's research-data destination.

```python
symbols = ["AAPL"]
start = datetime(2025, 1, 2, tzinfo=timezone.utc)
end = datetime(2025, 1, 3, tzinfo=timezone.utc)
output_path = None
```

## Fetch and Save Alternative Data

This cell fetches and saves the requested news articles in one call.

```python
saved_path = save_alpaca_news(
    symbols=symbols,
    start=start,
    end=end,
    output_path=output_path,
)
display(saved_path)
```

## Inspect Saved Data

The saved parquet file contains article timestamps, symbols, and text fields for sentiment scoring.

```python
alternative_data = pd.read_parquet(saved_path)
display(alternative_data.head())
alternative_data.shape
```
