# Fetch Fundamental Data

## Overview

This notebook saves Financial Modeling Prep statements through FinanceToolkit for fundamental feature construction.
- Problem: fundamental features need standardized statements and market-aware ratios.
- Approach: fetch quarterly income, balance-sheet, and cash-flow statements with FinanceToolkit.
- Configuration: set `FINANCIAL_MODELING_PREP_API_KEY` in `.env`.

```python
from datetime import datetime, timezone

import pandas as pd
from IPython.display import display

from src.data_preprocessing.fetch_fundamental_data import save_fmp_financial_statements
```

## Configure Request

This cell selects the tickers and financial-statement interval to fetch.

```python
symbols = ["AAPL"]
start = datetime(2025, 1, 1, tzinfo=timezone.utc)
end = datetime(2025, 12, 31, tzinfo=timezone.utc)
output_path = None
```

## Fetch and Save Fundamental Data

This cell fetches quarterly Financial Modeling Prep statements and saves them in one call.

```python
saved_path = save_fmp_financial_statements(
    symbols=symbols,
    start=start,
    end=end,
    output_path=output_path,
)
display(saved_path)
```

## Inspect Saved Data

The saved parquet file contains standardized FinanceToolkit statement rows and periods.

```python
fundamental_data = pd.read_parquet(saved_path)
display(fundamental_data.head())
fundamental_data.shape
```
