# Fetch Fundamental Data

## Overview

This notebook saves SEC 10-K and 10-Q financial statements for fundamental feature construction.
- Problem: fundamental features require accounting statements available as of their filing dates.
- Approach: fetch EdgarTools financial statements and retain filing metadata in parquet.
- Fundamental Data: It supports value, quality, profitability, growth, investment, leverage, efficiency, cash-flow, shareholder-yield, and size features.

```python
from datetime import datetime, timezone

import pandas as pd
from IPython.display import display

from src.data_preprocessing.fetch_fundamental_data import save_edgar_financial_statements
```

## Configure Request

This cell selects the companies and filing-date interval to fetch.
- `symbols` selects SEC registrants by ticker.
- `start` and `end` define the inclusive filing-date interval.
- `output_path=None` uses the module's research-data destination.

```python
symbols = ["AAPL"]
start = datetime(2025, 1, 1, tzinfo=timezone.utc)
end = datetime(2025, 12, 31, tzinfo=timezone.utc)
output_path = None
```

## Fetch and Save Fundamental Data

This cell fetches and saves the requested 10-K and 10-Q financial statements in one call.

```python
saved_path = save_edgar_financial_statements(
    symbols=symbols,
    start=start,
    end=end,
    output_path=output_path,
)
display(saved_path)
```

## Inspect Saved Data

The saved parquet file preserves standardized statement rows, periods, and filing metadata.

```python
fundamental_data = pd.read_parquet(saved_path)
display(fundamental_data.head())
fundamental_data.shape
```
