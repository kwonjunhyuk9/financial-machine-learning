# Fundamental Data

## Overview

This notebook demonstrates the public functions in `fundamental_data` .
- Problem: fundamental features need standardized fundamental data.
- Approach: fetch standardized fundamental data from `FMP`.

```python
from datetime import datetime, timezone

import pandas as pd
from IPython.display import display

from src.data_preprocessing.fundamental_data import save_fmp_financial_statements
```

## Observed Data

This cell defines the request for observed quarterly financial statements.
- `symbols` contains the instruments to fetch.
- `start` contains the inclusive UTC interval start.
- `end` contains the inclusive UTC interval end.
- `output_path` is `None`, so the dataset is saved at `data/research_data/fundamental/data/`.

```python
symbols = ["AAPL"]
start = datetime(2025, 1, 1, tzinfo=timezone.utc)
end = datetime(2025, 12, 31, tzinfo=timezone.utc)
output_path = None
```

## Fundamental Data Output

This cell fetches and saves quarterly financial statements, then reports the normalized fundamental-data fields.
- `statement_type` identifies `income_statement`, `balance_sheet`, or `cash_flow_statement`.
- `index` identifies the financial-statement line item.
- `2025Q1` contains the first-quarter reported value for that line item.
- `2025Q2` contains the second-quarter reported value for that line item.
- `2025Q3` contains the third-quarter reported value for that line item.
- `2025Q4` contains the fourth-quarter reported value for that line item.

```python
saved_path = save_fmp_financial_statements(
    symbols=symbols,
    start=start,
    end=end,
    output_path=output_path,
)

fundamental_data = pd.read_parquet(saved_path)
display(fundamental_data.head())
fundamental_data.shape
```
