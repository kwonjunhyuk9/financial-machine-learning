## Overview

This notebook demonstrates the public market-price operation in `kraken_client` using one configured Kraken symbol.
- Problem: the live runtime needs a current market price before it can validate a signal or submit an order.
- Approach: construct `KrakenClient` from the shared configuration and fetch the public last-traded price for its configured symbol.
- Market Price: It requests one public Kraken ticker and displays the normalized price used by the runtime.


```python
from IPython.display import display
```

## Market Price

This cell makes a read-only public network request.
- It does not submit, cancel, or synchronize an order.
- Kraken availability and network access are required to run this cell.
- The returned price is a point-in-time observation, not a historical price series.


```python
config = load_live_trading_config()
client = KrakenClient(config)
last_price = client.fetch_market_price(config.symbol)

display({"symbol": config.symbol, "last_price": last_price})
```
